# Test

```python
hw = HWSystem()

a = hw.wire("a", 3)
r = hw.wire("r", 3)
p = hw.wire("p", 3)

vcc = hw.wire('one', 1)
Constant(hw, 'one', 1, vcc)

Add(hw, "add", a, p, r)
Reg(hw, "reg", d=r, enable=vcc, q=p)

Constant(hw, "a", 1, a)

wvf = Waveform(hw, "wvf", [a,p,r])


hw.getSimulator().clk(20)

wvf.draw_wavedrom()
```
